## Inference

```python
from ultralytics import YOLO

model = YOLO(
    r"D:\Automatic ANPR\Self_Code\runs\detect\lp_yolov8s_stage2_final3\weights\best.pt"
)

metrics = model.val(
    data=r"D:\Automatic ANPR\Self_Code\dataset_preprocessed\data.yaml",
    imgsz=640,
    conf=0.001,   # low threshold for fair evaluation
    iou=0.6,      # standard IoU
    plots=True    # saves plots
)


print("mAP@0.5      :", metrics.box.map50)
print("mAP@0.5:0.95 :", metrics.box.map)
print("Precision    :", metrics.box.mp)
print("Recall       :", metrics.box.mr)
precision = metrics.box.mp
recall = metrics.box.mr

f1 = 2 * (precision * recall) / (precision + recall + 1e-6)
print("F1-score     :", f1)
print("Per-class AP:", metrics.box.ap)
```
