```python
import cv2
import os
import numpy as np

INPUT_DIR = r"C:\Users\100ra\OneDrive\Desktop\Automatic ANPR\Self_Code\dataset_original\images"
OUTPUT_DIR = r"C:\Users\100ra\OneDrive\Desktop\Automatic ANPR\Self_Code\dataset_preprocessed\images"

def preprocess(img):
    # Resize
    img = cv2.resize(img, (640, 640))

    # Convert to LAB for CLAHE
    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    l, a, b = cv2.split(lab)

    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    cl = clahe.apply(l)

    limg = cv2.merge((cl,a,b))
    img = cv2.cvtColor(limg, cv2.COLOR_LAB2BGR)

    # Gaussian blur (mild)
    img = cv2.GaussianBlur(img, (3,3), 0)

    # Sharpen
    kernel = np.array([[0,-1,0], [-1,5,-1], [0,-1,0]])
    img = cv2.filter2D(img, -1, kernel)

    return img

for split in ["train", "val"]:
    os.makedirs(os.path.join(OUTPUT_DIR, split), exist_ok=True)
    src = os.path.join(INPUT_DIR, split)

    for img_name in os.listdir(src):
        img_path = os.path.join(src, img_name)
        img = cv2.imread(img_path)

        if img is None:
            continue

        out = preprocess(img)
        cv2.imwrite(os.path.join(OUTPUT_DIR, split, img_name), out)

print("✅ Preprocessing complete")
```
